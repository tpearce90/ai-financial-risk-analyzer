```python
import pandas as pd
import matplotlib.pyplot as plt

prices = pd.read_csv(
    "../data/raw/prices.csv",
    index_col=0,
    parse_dates=True
)

prices.head()
```

```python
prices.plot(
    figsize=(12, 6),
    title="Asset Prices Over Time"
)

plt.xlabel("Date")
plt.ylabel("Price")
plt.show()
```

```python
normalized_prices = prices / prices.iloc[0]

normalized_prices.head()
```

```python
normalized_prices.plot(
    figsize=(12, 6),
    title="Normalized Asset Performance"
)

plt.xlabel("Date")
plt.ylabel("Normalized Growth")
plt.show()
```
